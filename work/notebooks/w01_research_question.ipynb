## 1. My lane (or freestyle) and why

I choose the Ranking Signal Analysis lane. It is a strong fit because the starter data already contains many observable content and search signals, while the review budget is limited. I want to study which signals are associated with decline or opportunity so I can rank pages for review, not claim that I have proved a causal effect.

```python
from pathlib import Path
import pandas as pd


def find_dataset() -> Path:
    candidates = [
        Path.cwd() / "data" / "raw" / "content_refresh_anonymized.csv",
        Path.cwd().parent / "data" / "raw" / "content_refresh_anonymized.csv",
        Path.cwd().parent.parent / "data" / "raw" / "content_refresh_anonymized.csv",
        Path("data/raw/content_refresh_anonymized.csv"),
    ]
    for candidate in candidates:
        if candidate.exists():
            return candidate
    raise FileNotFoundError("Could not find the starter dataset.")

path = find_dataset()
df = pd.read_csv(path)
print("Starter rows:", len(df))
print("Relevant columns present:", [c for c in ["impressions_90d", "sessions_90d", "ctr", "avg_position", "trend_direction"] if c in df.columns])
```

## 2. The question: decision, action, cost of a wrong call

My decision question is: which content pages should a reviewer inspect first? The action is a manual refresh review or a deeper content audit. The cost of a wrong recommendation is wasted reviewer time and a missed opportunity on a page that might actually deserve attention. Data and ML help because there are thousands of pages but only limited review capacity, so a ranking approach can surface the most promising candidates from observed signals.

```python
import pandas as pd

# Use the same dataset loaded above to show a concrete decision problem.
df["is_declining"] = df["trend_direction"].fillna("").str.lower().eq("down")
df["high_demand"] = df["impressions_90d"] >= 1000
high_demand_declining = df[df["is_declining"] & df["high_demand"]]

print("High-demand declining pages:", len(high_demand_declining))
print("Median CTR among them:", round(high_demand_declining["ctr"].median(), 3))
print("Median avg position among them:", round(high_demand_declining["avg_position"].median(), 2))
```

## 3. Quick look at the data (2-3 real numbers)

The starter dataset already makes this lane look worth the next 7 weeks. The code below shows three concrete numbers from the real starter CSV: the size of the sample, the share of pages that look declining, and the number of pages with strong demand.

```python
# Three concrete starter-dataset numbers that support the lane choice.
df["is_declining"] = df["trend_direction"].fillna("").str.lower().eq("down")

print("rows:", len(df))
print("declining share:", round(df["is_declining"].mean() * 100, 1), "%")
print("pages with >1000 impressions:", int((df["impressions_90d"] > 1000).sum()))
```
